```python
# ARITHMETIC OPR.
a = 20
b = 5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(type(a/b))
print(a//b)
print(type(a//b))

A = 12
B = 5
print(A%B)
print(A**B)

x = 20
y = 6
print(x // y)
```

```python
# ASSIGNMENT OPR.

x = 10
print(x)

x = 20
print(x)

a = 10
a += 5    # a = a + 5
print(a)

marks = 80
marks += 10
print(marks)

x = 50
x -= 20
print(x)

salary = 2000
salary *= 2
print(salary)

x = 50
x /= 5
print(x)

x = 17
x %= 5
print(x)

x = 5
x **= 2
print(x)
```

```python
score = 100

score += 20
print(score)

score -= 10
print(score)

score *= 2
print(score)

score /= 5
print(score)
```
