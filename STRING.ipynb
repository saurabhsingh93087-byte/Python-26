```python
# STRING FUNCTIONS.

"""
                                      Important String Methods for Exams & Interviews
                                      
                                                      Method	Use
                                                      upper()	Uppercase
                                                      lower()	Lowercase
                                                      capitalize()	First letter capital
                                                      title()	   Every word capital
                                                      swapcase()	Change case
                                                      strip()	Remove spaces
                                                      replace()	Replace text
                                                      find()	Find position (-1 if not found)
                                                      index()	Find position (Error if not found)
                                                      count()	Count occurrences
                                                      startswith()	Check starting text
                                                      endswith()	Check ending text
                                                     * split()	String → List
                                                     * join()	List → String
                                                      isalpha()	Only letters
                                                      isdigit()	Only digits
                                                      isalnum()	Letters + digits
                                                      islower()	All lowercase
                                                      isupper()	All uppercase
                                                      isspace()	Only spaces
                                                      center()	Center align
                                                      ljust()	Left align
                                                      rjust()	Right align
                                                      zfill()	Add leading zeros                                                                    """


#1.upper() -> Uppercase
text = "python programming"
print(text.upper())

#2.lower() -> Lowercase
name = "SAURABH RAJPUT"
print(name.lower())

#3.capitalize()	-> First letter capital
text = "python programming"
print(text.capitalize())

name = "saurabh rajput"
print(name.capitalize())

#4.title() -> Every 1st letter of word capital.
text = "python programming language"
print(text.title())
 
#5.swapcase() -> Uppercase ko lowercase aur lowercase ko uppercase bana deta hai.
name = "saURabhSINgh"
print(name.swapcase())

```

```python
#6.strip() ->String ke starting aur ending ke extra spaces remove karta hai.
name = "      saurabh singh      "
print(name.strip())

#7. lstrip() ->Sirf left side ke spaces remove karta hai.
text = "   Python"
print(text.lstrip())
#8.rstrip -> only right side spaces remove.
name = "   saurabh       "
print(name.rstrip())

#9.replace() ->Ek word ya character ko dusre se replace karta hai. 
fruit = "I likes Banana."
print(fruit.replace("Banana","Mango"))

#10. find() -> Character ya word ki first position batata hai,  Agar nahi mila to -1 return karta hai.
name = "saurabh"
print(name.find("r"))
print(name.find("n"))
print(name.find("a"))

#11. index() -> Ye bhi position batata hai. Difference:Agar character nahi mila to ValueError deta hai.
name = "saurabh"
print(name.index("r"))
print(name.index("n"))
```

```python
#12.count() -> Koi character kitni baar aaya hai.
text = "banana"
print(text.count("a"))

#13. startswith() & endswith() -> Check karta hai string kis word se start or end  hoti hai.
text = "Python Programming"
print(text.startswith("Python"))
print(text.endswith("Python"))


text = "Python Programming"
print(text.endswith("Programming"))

#15. split() -> Ek string ko list mein convert karta hai.
text= "I love Quad Ai"
print(text.split())

fruit = "Apple,Banana,Mango"
print(fruit.split(","))

#16. join() ->List ko string mein convert karta ha.
fruits = ["Seb","Kela","Aam"]
print("-".join(fruits))
```

```python
#17. isalpha()  ->Sirf letters hain ya nahi. 
print("Python".isalpha())
print("Python123".isalpha())

#18. isdigit()  ->Sirf digits hain ya nahi.
print("12345".isdigit())
print("123A".isdigit())

#21. islower()  ->Sab lowercase hain ya nahi.
print("pythOn".islower())

#22.isupper()  -> sab uppercase hai ya nahi.
print("SAURABH".isupper())

#24. center()  ->String ko center mein align karta hai.
text = "Python"
print(text.center(20))
print(text.center(50,"*"))

#25. ljust()  ->Left align karta hai.& rjust() ->Right align karta hai.
text = "Python"
print(text.ljust(15,"-"))
print(text.rjust(20,"-"))

#26. zfill()  ->Left side mein zeros add karta hai.
num = "25"
print(num.zfill(5))

```
