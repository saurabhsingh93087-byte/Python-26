```python
# calculator


print("------ Calculator ------")

choice = input("Enter your choice (+,-,*,/): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == "+":
    print("Answer =", num1 + num2)

elif choice == "-":
    print("Answer =", num1 - num2)

elif choice == "*":
    print("Answer =", num1 * num2)

elif choice == "/":
    if num2 != 0:
        print("Answer =", num1 / num2)
    else:
        print("Division by zero is not possible.")

else:
    print("Invalid Choice")




```
