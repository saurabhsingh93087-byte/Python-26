```python
# Type Casting (Data Type Conversion).
```

```python
# Implicit Type Casting.
a = 10
b = 30.5
sum = a+b
print(sum)
print(type(sum))
```

```python
# Explicit Type Casting.
age = "18"
print(type(age))

age = int(age)

print(age)
print(type(age))
```
