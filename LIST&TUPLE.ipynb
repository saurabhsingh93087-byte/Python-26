```python

fruits = ["apple","banana","mango","licchi"]
print(fruits)

print(fruits[0])
print(fruits[1])
print(fruits[3])
print(fruits[-2])

```

```python
#changing items;
fruits = ["Apple", "Banana", "Mango", "dragon"]
fruits[1] = "Orange"
fruits[3] = "Guava"
print(fruits)
```

```python
#adding items(end);
animals = ["horse","cow","tiger"]
animals.append("dog")
print(animals)

#add items(at any);
animals = ["horse","cow","tiger"]
animals.insert(2,"cat")
print(animals)
animals.insert(0,"rat")
print(animals)
```

```python
#remove items;
colours = ["white","black","yellow","grey","pink"]
colours.remove("black")
print(colours)

#Removes by index.
colours = ["white","black","yellow","grey","pink"]
colours.pop(3)
print(colours)

#delete num;
numbers = [2,4,6,8,10,12]
del numbers[2]
print(numbers)
```
